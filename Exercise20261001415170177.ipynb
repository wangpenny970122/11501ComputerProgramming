1.將十進位小數 0.1 轉換成二進位：
因此可得到：
0.1₁₀ = 0.0001100110011001100110011...₂
其中：
0011
會不斷重複出現，因此是無限循環二進位小數。

2.Python 的 float 採用 IEEE 754 Double Precision（雙精度浮點數）。
結構:
1 bit 符號位
11 bits 指數位
52 bits 尾數位

````python
zodiac = [
    "monkey",
    "rooster",
    "dog",
    "pig",
    "rat",
    "ox",
    "tiger",
    "rabbit",
    "dragon",
    "snake",
    "horse",
    "sheep"
]

year = int(input("請輸入年份："))

# 若輸入為民國年，轉換成西元年
if year < 1912:
    year = year + 1911

print("生肖為：", zodiac[year % 12])
````

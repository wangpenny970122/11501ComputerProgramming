### Part 1：完成函數

請閱讀以下函數，理解各部分的功能。

```python
def analyze_scores(*scores, **options):
    """
    分析學生分數。

    Parameters:
        *scores: 接收不定數量的分數
        **options: 接收選擇性設定，例如 multiplier 和 bonus

    Returns:
        一個字典，包含調整後的分數、平方、總分及平均分數。
    """
    multiplier = options.get("multiplier", 1)
    bonus = options.get("bonus", 0)

    adjusted = [
        score * multiplier + bonus
        for score in scores
    ]

    squared = [
        score ** 2
        for score in adjusted
    ]

    total = sum(adjusted)
    average = total / len(adjusted) if adjusted else 0

    return {
        "adjusted_scores": adjusted,
        "squared_scores": squared,
        "total": total,
        "average": average
    }
```

### part4.
1.*scores 用來接收不定數量的位置參數（Positional Arguments）。
scores_list 是一個關鍵字參數（Keyword Argument），通常接收一個串列（list）。
2.
因為沒有傳入任何位置參數。雖然 scores_list 被接收到，但原本函數並沒有將：options["scores_list"]取出來參與計算，因此 adjusted 會是空串列，無法得到預期結果。


````python
3-1
def analyze_scores(*scores, **options):
    multiplier = options.get("multiplier", 1)
    bonus = options.get("bonus", 0)

    if not scores:
        scores = options.get("scores_list", [])

    adjusted = [
        score * multiplier + bonus
        for score in scores
    ]

    squared = [
        score ** 2
        for score in adjusted
    ]

    total = sum(adjusted)
    average = total / len(adjusted) if adjusted else 0

    max_score = max(adjusted) if adjusted else 0

    return {
        "adjusted_scores": adjusted,
        "squared_scores": squared,
        "total": total,
        "average": average,
        "max_score": max_score
    }
result = analyze_scores(
    60, 70, 80,
    multiplier=1.1,
    bonus=5
)

print(result)
````

四、總結

在 Python 中，*、** 與 return 都是非常重要的語法，但用途各不相同。

1.* 的用途

* 主要有兩種常見用途：
(1) 接收不定數量的位置參數
(2) 解包（Unpacking）串列或元組

2. ** 的用途
(1) 接收不定數量的關鍵字參數
(2) 解包字典

3. return 的用途
return 用來將函數執行結果回傳給呼叫者。

--- 同一個符號在不同位置的不同意義

-* 的例子

函數定義時：
def func(*nums):
    print(nums)
接收多個位置參數

函數呼叫時：
nums = [1, 2, 3]
func(*nums)
將串列解包成多個位置參數

** 的例子

函數定義時：
def func(**data):
    print(data)
接收多個關鍵字參數

函數呼叫時：
data = {
    "name": "Amy",
    "age": 20
}
func(**data)
將字典解包成多個關鍵字參數

---
*：接收或解包位置參數。
**：接收或解包關鍵字參數。
return：將函數計算結果回傳給呼叫者，以便後續使用。
同一個 * 或 **，在「函數定義」與「函數呼叫」的位置不同，其代表的意義也不同。這些功能使 Python 的函數設計更有彈性，能夠處理各種不同形式的資料輸入。
